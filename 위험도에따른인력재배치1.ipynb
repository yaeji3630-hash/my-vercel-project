1-3. Google Drive 영구저장 파일 검수
- 저장한 파일이 실제 드라이브에 존재하는지 확인

```python
# @title
# =========================================================
# 1-3. Google Drive 영구저장 파일 검수
# =========================================================

from pathlib import Path
import pandas as pd

RAW_DIR = Path(
    "/content/drive/MyDrive/대전_물류창고_폭염위험도/01_원본데이터"
)

records = []

for file_path in sorted(RAW_DIR.iterdir()):

    if file_path.is_file():

        size_bytes = file_path.stat().st_size

        records.append({
            "파일명": file_path.name,
            "파일크기_Bytes": size_bytes,
            "파일크기_KB": round(size_bytes / 1024, 2),
            "정상저장": "정상" if size_bytes > 0 else "확인필요"
        })

check_df = pd.DataFrame(records)

print("저장 파일 수:", len(check_df))
print("0 byte 파일 수:", (check_df["파일크기_Bytes"] == 0).sum())

display(check_df)
```

2-1. 핵심 원본파일 일괄 구조, 결측 확인
- 파일이 정상적으로 읽히는지
- 행/열 수

```python
# =========================================================
# 2-1. 핵심 원본자료 일괄 구조·결측 확인
# =========================================================

from pathlib import Path
import pandas as pd

RAW_DIR = Path(
    "/content/drive/MyDrive/대전_물류창고_폭염위험도/01_원본데이터"
)

# ---------------------------------------------------------
# 1. 핵심 파일 목록
# ---------------------------------------------------------

files_to_check = {
    "물류창고": RAW_DIR / "대전_물류창고_상세정보_20260721.xlsx",
    "택배물량": RAW_DIR / "과학기술정보통신부 우정사업본부_전국 우편번호(배달지 기준)별 택배물량_20231231.csv",
    "교통량": RAW_DIR / "한국도로공사_구간교통량(월별 및 차종별)_20241231.csv",
    "법정동코드": RAW_DIR / "국토교통부_법정동코드_20250805.csv",
}

for year in range(2021, 2027):
    files_to_check[f"시간기상_{year}"] = (
        RAW_DIR / f"대전_기상청_시간자료_{year}.csv"
    )

for year in range(2020, 2027):
    files_to_check[f"체감온도_{year}"] = (
        RAW_DIR / f"STCS_체감온도_{year}년도.xls"
    )


# ---------------------------------------------------------
# 2. 파일 자동 읽기 함수
# ---------------------------------------------------------

def read_file(path):

    suffix = path.suffix.lower()

    if suffix in [".xlsx", ".xls"]:
        return pd.read_excel(path)

    elif suffix == ".csv":

        # 공공데이터에서 자주 쓰이는 인코딩 순차 시도
        for enc in ["utf-8-sig", "cp949", "euc-kr", "utf-8"]:
            try:
                return pd.read_csv(path, encoding=enc)
            except:
                continue

        raise ValueError(f"CSV 인코딩 확인 필요: {path.name}")

    else:
        raise ValueError(f"지원하지 않는 형식: {path.name}")


# ---------------------------------------------------------
# 3. 전체 파일 일괄 검수
# ---------------------------------------------------------

summary = []
loaded_data = {}

for name, path in files_to_check.items():

    try:
        df = read_file(path)

        loaded_data[name] = df

        summary.append({
            "데이터": name,
            "행수": len(df),
            "열수": len(df.columns),
            "결측총수": int(df.isna().sum().sum()),
            "중복행수": int(df.duplicated().sum()),
            "상태": "정상"
        })

    except Exception as e:

        summary.append({
            "데이터": name,
            "행수": None,
            "열수": None,
            "결측총수": None,
            "중복행수": None,
            "상태": f"오류: {str(e)[:80]}"
        })


summary_df = pd.DataFrame(summary)

print("[핵심 원본자료 통합 검수 결과]")
display(summary_df)

print("\n[각 데이터 열 이름]")

for name, df in loaded_data.items():

    print(f"\n▶ {name}")
    print(df.columns.tolist())
```

2-13. 11개 물류창고 격자의 전체 단기예보 수집
- TMP(기온), REH(습도)를 전체 예보기간에 대해 수집
- 전국 자료에서 필요한 11개 격자만 추출하여 Drive 저장

```python
# =========================================================
# 2-13. 11개 물류창고 격자의 전체 단기예보 수집
# =========================================================

from pathlib import Path
import pandas as pd
import numpy as np
import requests
import getpass
from datetime import datetime, timedelta
import time

# ---------------------------------------------------------
# 1. 폴더 및 격자정보 불러오기
# ---------------------------------------------------------

PROCESSED_DIR = Path(
    "/content/drive/MyDrive/대전_물류창고_폭염위험도/03_전처리데이터"
)

API_DIR = Path(
    "/content/drive/MyDrive/대전_물류창고_폭염위험도/02_API수집데이터"
)

warehouse = pd.read_csv(
    PROCESSED_DIR / "대전_물류창고_좌표_기상청격자.csv",
    encoding="utf-8-sig"
)

grids = (
    warehouse[["nx", "ny"]]
    .drop_duplicates()
    .sort_values(["nx", "ny"])
    .reset_index(drop=True)
)

print("조회 대상 고유 격자:", len(grids), "개")


# ---------------------------------------------------------
# 2. 기상청 API허브 인증키
# ---------------------------------------------------------

AUTH_KEY = getpass.getpass(
    "기상청 API허브 인증키를 입력하세요: "
)

url = (
    "https://apihub.kma.go.kr/api/typ01/"
    "cgi-bin/url/nph-dfs_shrt_grd"
)


# ---------------------------------------------------------
# 3. 최근 사용 가능한 단기예보 발표시간 결정
# ---------------------------------------------------------

now = datetime.now()

base_hours = [2, 5, 8, 11, 14, 17, 20, 23]

tmfc_dt = None

for h in reversed(base_hours):

    candidate = now.replace(
        hour=h,
        minute=0,
        second=0,
        microsecond=0
    )

    # 자료 생성 지연 고려
    if candidate <= now - timedelta(minutes=20):
        tmfc_dt = candidate
        break

if tmfc_dt is None:

    tmfc_dt = (
        now - timedelta(days=1)
    ).replace(
        hour=23,
        minute=0,
        second=0,
        microsecond=0
    )

tmfc = tmfc_dt.strftime("%Y%m%d%H")

print("사용 발표시간:", tmfc)


# ---------------------------------------------------------
# 4. 전체 예보시간 생성
# ---------------------------------------------------------

# 02~14시 발표 → 글피 00시까지
# 17~23시 발표 → 그글피 00시까지

if tmfc_dt.hour in [17, 20, 23]:

    end_dt = (
        tmfc_dt.replace(
            hour=0,
            minute=0,
            second=0,
            microsecond=0
        )
        + timedelta(days=4)
    )

else:

    end_dt = (
        tmfc_dt.replace(
            hour=0,
            minute=0,
            second=0,
            microsecond=0
        )
        + timedelta(days=3)
    )

start_dt = (
    tmfc_dt + timedelta(hours=1)
).replace(
    minute=0,
    second=0,
    microsecond=0
)

forecast_times = pd.date_range(
    start=start_dt,
    end=end_dt,
    freq="h"
).to_pydatetime().tolist()

print(
    "예보 수집기간:",
    forecast_times[0].strftime("%Y-%m-%d %H:%M"),
    "~",
    forecast_times[-1].strftime("%Y-%m-%d %H:%M")
)

print(
    "예보시각 수:",
    len(forecast_times)
)


# ---------------------------------------------------------
# 5. TMP / REH 전체 예보 수집
# ---------------------------------------------------------

NX_FULL = 149
NY_FULL = 253
GRID_SIZE = NX_FULL * NY_FULL

all_rows = []

successful_times = 0
failed_times = []

for n, fcst_dt in enumerate(
    forecast_times,
    start=1
):

    tmef = fcst_dt.strftime("%Y%m%d%H")

    arrays = {}

    # TMP와 REH 각각 수집
    for var in ["TMP", "REH"]:

        try:

            response = requests.get(
                url,
                params={
                    "tmfc": tmfc,
                    "tmef": tmef,
                    "vars": var,
                    "authKey": AUTH_KEY
                },
                timeout=30
            )

            if response.status_code != 200:
                continue

            # 2-12에서 검증한 공식 파싱 방식
            arr = np.fromstring(
                string=response.content,
                dtype=np.float32,
                count=GRID_SIZE,
                sep=","
            )

            if arr.size != GRID_SIZE:
                continue

            arrays[var] = arr.reshape(
                NY_FULL,
                NX_FULL
            )

        except Exception:
            continue


    # TMP와 REH가 모두 있는 시각만 채택
    if (
        "TMP" not in arrays
        or "REH" not in arrays
    ):

        failed_times.append(tmef)
        continue


    # -----------------------------------------------------
    # 필요한 11개 격자만 추출
    # -----------------------------------------------------

    for _, g in grids.iterrows():

        nx = int(g["nx"])
        ny = int(g["ny"])

        tmp = float(
            arrays["TMP"][ny - 1, nx - 1]
        )

        reh = float(
            arrays["REH"][ny - 1, nx - 1]
        )

        # 기상청 결측값 처리
        if tmp <= -99:
            tmp = np.nan

        if reh <= -99:
            reh = np.nan

        all_rows.append({
            "발표시간": tmfc,
            "예보일자": fcst_dt.strftime("%Y%m%d"),
            "예보시각": fcst_dt.strftime("%H00"),
            "예보일시": fcst_dt,
            "nx": nx,
            "ny": ny,
            "TMP": tmp,
            "REH": reh
        })

    successful_times += 1


    # -----------------------------------------------------
    # 진행률 표시
    # -----------------------------------------------------

    if (
        n % 10 == 0
        or n == len(forecast_times)
    ):

        print(
            f"[진행] {n}/{len(forecast_times)} "
            f"예보시각 처리"
        )

    time.sleep(0.03)


# ---------------------------------------------------------
# 6. 데이터프레임 생성
# ---------------------------------------------------------

forecast = pd.DataFrame(all_rows)

if forecast.empty:

    raise RuntimeError(
        "단기예보 자료가 수집되지 않았습니다."
    )


# ---------------------------------------------------------
# 7. 최소 검수
# ---------------------------------------------------------

print("\n[단기예보 최종 수집 결과]")

print(
    "정상 수집 예보시각:",
    successful_times,
    "/",
    len(forecast_times)
)

print(
    "수집 행 수:",
    len(forecast)
)

print(
    "수집 격자 수:",
    forecast[["nx", "ny"]]
    .drop_duplicates()
    .shape[0]
)

print(
    "TMP 결측:",
    forecast["TMP"].isna().sum()
)

print(
    "REH 결측:",
    forecast["REH"].isna().sum()
)

print(
    "중복:",
    forecast.duplicated(
        subset=["예보일시", "nx", "ny"]
    ).sum()
)

print(
    "예보기간:",
    forecast["예보일시"].min(),
    "~",
    forecast["예보일시"].max()
)

if failed_times:
    print(
        "\n수집 실패 예보시각 수:",
        len(failed_times)
    )
    print(
        "실패 시각:",
        failed_times
    )
else:
    print(
        "\n모든 예보시각 정상 수집"
    )


display(
    forecast.head(10)
)


# ---------------------------------------------------------
# 8. Google Drive 저장
# ---------------------------------------------------------

save_path = (
    API_DIR /
    f"기상청_API허브_단기예보_11개격자_{tmfc}.csv"
)

forecast.to_csv(
    save_path,
    index=False,
    encoding="utf-8-sig"
)

print("\n저장 완료:")
print(save_path)
```

2-16. 기상청 공식 여름철 체감온도 재산출

```python
# =========================================================
# 2-16. 기상청 공식 여름철 체감온도 재산출
# =========================================================

from pathlib import Path
import pandas as pd
import numpy as np

PROCESSED_DIR = Path(
    "/content/drive/MyDrive/대전_물류창고_폭염위험도/03_전처리데이터"
)

# ---------------------------------------------------------
# 1. 2-14에서 만든 창고별 시간별 기온·습도 예보 불러오기
#
# 2-15에서 잘못 계산한 Heat Index 파일을 사용하지 않고,
# 원래 TMP·REH가 들어있는 2-14 파일에서 다시 계산
# ---------------------------------------------------------

path = (
    PROCESSED_DIR /
    "대전_43개물류창고_시간별단기예보.csv"
)

df = pd.read_csv(
    path,
    encoding="utf-8-sig"
)

df["예보일시"] = pd.to_datetime(
    df["예보일시"]
)


# ---------------------------------------------------------
# 2. Stull 추정식으로 습구온도(Tw) 계산
#
# Ta : 기온(℃)
# RH : 상대습도(%)
# np.arctan()은 라디안 기준이며 공식 적용에 그대로 사용
# ---------------------------------------------------------

def calc_wet_bulb_temperature(Ta, RH):

    Tw = (
        Ta
        * np.arctan(
            0.151977
            * np.sqrt(RH + 8.313659)
        )
        + np.arctan(Ta + RH)
        - np.arctan(RH - 1.67633)
        + 0.00391838
        * (RH ** 1.5)
        * np.arctan(0.023101 * RH)
        - 4.686035
    )

    return Tw


# ---------------------------------------------------------
# 3. 기상청 여름철 체감온도 계산
#
# ST =
# -0.2442
# + 0.55399*Tw
# + 0.45535*Ta
# - 0.0022*Tw^2
# + 0.00278*Tw*Ta
# + 3.0
# ---------------------------------------------------------

df["습구온도"] = calc_wet_bulb_temperature(
    df["TMP"],
    df["REH"]
)

df["체감온도"] = (
    -0.2442
    + 0.55399 * df["습구온도"]
    + 0.45535 * df["TMP"]
    - 0.0022 * (df["습구온도"] ** 2)
    + 0.00278 * df["습구온도"] * df["TMP"]
    + 3.0
)


# ---------------------------------------------------------
# 4. 소수점 정리
# ---------------------------------------------------------

df["습구온도"] = (
    df["습구온도"]
    .round(2)
)

df["체감온도"] = (
    df["체감온도"]
    .round(1)
)


# ---------------------------------------------------------
# 5. 최소 검수
# ---------------------------------------------------------

print("[기상청 공식 체감온도 산출 결과]")

print(
    "전체 행 수:",
    len(df)
)

print(
    "체감온도 결측:",
    df["체감온도"].isna().sum()
)

print(
    "체감온도 최소:",
    df["체감온도"].min(),
    "℃"
)

print(
    "체감온도 최대:",
    df["체감온도"].max(),
    "℃"
)

print(
    "체감온도 평균:",
    round(df["체감온도"].mean(), 1),
    "℃"
)


# ---------------------------------------------------------
# 6. 상위 체감온도 확인
# ---------------------------------------------------------

print("\n체감온도 상위 10개:")

display(
    df[
        [
            "WARE_NO",
            "COMPANY_NAME",
            "예보일시",
            "TMP",
            "REH",
            "습구온도",
            "체감온도"
        ]
    ]
    .sort_values(
        "체감온도",
        ascending=False
    )
    .head(10)
)


# ---------------------------------------------------------
# 7. 검증용 예시
#    33℃ / 70% → 약 34.3℃가 나오는지 확인
# ---------------------------------------------------------

test_Ta = 33
test_RH = 70

test_Tw = calc_wet_bulb_temperature(
    test_Ta,
    test_RH
)

test_ST = (
    -0.2442
    + 0.55399 * test_Tw
    + 0.45535 * test_Ta
    - 0.0022 * (test_Tw ** 2)
    + 0.00278 * test_Tw * test_Ta
    + 3.0
)

print("\n[공식 검증 예시]")
print("기온:", test_Ta, "℃")
print("습도:", test_RH, "%")
print("습구온도:", round(test_Tw, 2), "℃")
print("체감온도:", round(test_ST, 1), "℃")


# ---------------------------------------------------------
# 8. 최종 저장
# ---------------------------------------------------------

save_path = (
    PROCESSED_DIR /
    "대전_43개물류창고_시간별체감온도예보_기상청공식.csv"
)

df.to_csv(
    save_path,
    index=False,
    encoding="utf-8-sig"
)

print("\n저장 완료:")
print(save_path)
```

```python
# =========================================================
# 2-17. 대전 중기기온 예보 4~10일 수집
# =========================================================

from pathlib import Path
import pandas as pd
import requests
import getpass
from datetime import datetime, timedelta

API_DIR = Path(
    "/content/drive/MyDrive/대전_물류창고_폭염위험도/02_API수집데이터"
)

API_DIR.mkdir(parents=True, exist_ok=True)

# ---------------------------------------------------------
# 1. 기상청 API허브 인증키 입력
# ---------------------------------------------------------

AUTH_KEY = getpass.getpass(
    "기상청 API허브 인증키를 입력하세요: "
)

# ---------------------------------------------------------
# 2. 최근 사용 가능한 중기예보 발표시각 결정
#
# 중기예보 발표: 매일 06시 / 18시
# 최근 24시간 자료만 조회 가능
# ---------------------------------------------------------

now = datetime.now()

if now.hour >= 19:
    tmfc_dt = now.replace(
        hour=18,
        minute=0,
        second=0,
        microsecond=0
    )

elif now.hour >= 7:
    tmfc_dt = now.replace(
        hour=6,
        minute=0,
        second=0,
        microsecond=0
    )

else:
    tmfc_dt = (
        now - timedelta(days=1)
    ).replace(
        hour=18,
        minute=0,
        second=0,
        microsecond=0
    )

tmFc = tmfc_dt.strftime("%Y%m%d%H00")

print("사용 발표시각:", tmFc)


# ---------------------------------------------------------
# 3. 대전 중기기온 API 호출
#
# 대전 중기기온 예보구역코드: 11C20401
# ---------------------------------------------------------

url = (
    "https://apihub.kma.go.kr/api/typ02/openApi/"
    "MidFcstInfoService/getMidTa"
)

params = {
    "pageNo": 1,
    "numOfRows": 10,
    "dataType": "JSON",
    "regId": "11C20401",
    "tmFc": tmFc,
    "authKey": AUTH_KEY
}

response = requests.get(
    url,
    params=params,
    timeout=30
)

print("HTTP 상태:", response.status_code)

response.raise_for_status()

data = response.json()


# ---------------------------------------------------------
# 4. API 응답 확인
# ---------------------------------------------------------

header = (
    data
    .get("response", {})
    .get("header", {})
)

result_code = header.get("resultCode")
result_msg = header.get("resultMsg")

print(
    "API 결과:",
    result_code,
    "/",
    result_msg
)

if result_code != "00":
    raise RuntimeError(
        f"중기예보 API 오류: "
        f"{result_code} / {result_msg}"
    )


# ---------------------------------------------------------
# 5. 첫 번째 결과 추출
# ---------------------------------------------------------

items = (
    data["response"]["body"]
    ["items"]["item"]
)

if not items:
    raise RuntimeError(
        "중기기온 자료가 없습니다."
    )

item = items[0]


# ---------------------------------------------------------
# 6. 4~10일 예보를 날짜별 행으로 변환
# ---------------------------------------------------------

rows = []

for day in range(4, 11):

    forecast_date = (
        tmfc_dt.date()
        + timedelta(days=day)
    )

    rows.append({
        "발표시각": tmFc,
        "예보일": forecast_date,

        "최저기온": item.get(
            f"taMin{day}"
        ),

        "최저기온_하한": item.get(
            f"taMin{day}Low"
        ),

        "최저기온_상한": item.get(
            f"taMin{day}High"
        ),

        "최고기온": item.get(
            f"taMax{day}"
        ),

        "최고기온_하한": item.get(
            f"taMax{day}Low"
        ),

        "최고기온_상한": item.get(
            f"taMax{day}High"
        )
    })


mid_temp = pd.DataFrame(rows)


# ---------------------------------------------------------
# 7. 숫자형 변환
# ---------------------------------------------------------

temp_cols = [
    "최저기온",
    "최저기온_하한",
    "최저기온_상한",
    "최고기온",
    "최고기온_하한",
    "최고기온_상한"
]

for col in temp_cols:

    mid_temp[col] = pd.to_numeric(
        mid_temp[col],
        errors="coerce"
    )


# ---------------------------------------------------------
# 8. 최소 검수
# ---------------------------------------------------------

print("\n[대전 중기기온 수집 결과]")

print(
    "수집 일수:",
    len(mid_temp)
)

print(
    "예보기간:",
    mid_temp["예보일"].min(),
    "~",
    mid_temp["예보일"].max()
)

print(
    "최저기온 결측:",
    mid_temp["최저기온"].isna().sum()
)

print(
    "최고기온 결측:",
    mid_temp["최고기온"].isna().sum()
)

print(
    "기간 내 최고 예상기온:",
    mid_temp["최고기온"].max(),
    "℃"
)

display(mid_temp)


# ---------------------------------------------------------
# 9. 저장
# ---------------------------------------------------------

save_path = (
    API_DIR /
    f"기상청_API허브_대전_중기기온_{tmFc}.csv"
)

mid_temp.to_csv(
    save_path,
    index=False,
    encoding="utf-8-sig"
)

print("\n저장 완료:")
print(save_path)
```

```python
# =========================================================
# 3-1. 43개 물류창고 우편번호 확보 및 검수
# =========================================================

from pathlib import Path
import pandas as pd
import requests
import getpass
import re
import time

API_DIR = Path(
    "/content/drive/MyDrive/대전_물류창고_폭염위험도/02_API수집데이터"
)

PROCESSED_DIR = Path(
    "/content/drive/MyDrive/대전_물류창고_폭염위험도/03_전처리데이터"
)

# ---------------------------------------------------------
# 1. 최종 좌표 파일 불러오기
# ---------------------------------------------------------

warehouse_path = (
    API_DIR /
    "대전_물류창고_카카오좌표_최종.csv"
)

warehouse = pd.read_csv(
    warehouse_path,
    encoding="utf-8-sig",
    dtype=str
)

print("불러온 창고 수:", len(warehouse))
print("현재 열:", warehouse.columns.tolist())


# ---------------------------------------------------------
# 2. 기존 우편번호 열 확인
# ---------------------------------------------------------

possible_zip_cols = [
    "카카오_우편번호",
    "우편번호",
    "zone_no",
    "zoneNo"
]

existing_zip_col = next(
    (
        col for col in possible_zip_cols
        if col in warehouse.columns
    ),
    None
)

if existing_zip_col is None:
    warehouse["우편번호"] = pd.NA
else:
    warehouse["우편번호"] = warehouse[existing_zip_col]


# 우편번호를 숫자만 남긴 5자리 문자열로 정리
warehouse["우편번호"] = (
    warehouse["우편번호"]
    .astype("string")
    .str.replace(r"\D", "", regex=True)
    .str.zfill(5)
)

# 빈 값이 zfill 때문에 잘못 변환되는 것을 방지
invalid_zip = (
    warehouse["우편번호"].isna()
    | (warehouse["우편번호"].str.len() != 5)
    | (warehouse["우편번호"] == "00<NA>")
)

warehouse.loc[invalid_zip, "우편번호"] = pd.NA


# ---------------------------------------------------------
# 3. 주소 표준화 함수
#
# 예:
# 411번지 3호 → 411-3
# ---------------------------------------------------------

def normalize_lot_address(address):

    if pd.isna(address):
        return None

    address = str(address).strip()

    address = re.sub(
        r"(\d+)\s*번지\s*(\d+)\s*호",
        r"\1-\2",
        address
    )

    address = re.sub(
        r"(\d+)\s*번지",
        r"\1",
        address
    )

    address = re.sub(r"\s+", " ", address)

    return address.strip()


warehouse["우편번호검색주소"] = (
    warehouse["COMPANY_ADDRESS"]
    .apply(normalize_lot_address)
)


# ---------------------------------------------------------
# 4. 우편번호 미확보 창고만 카카오 API 재조회
# ---------------------------------------------------------

missing_before = warehouse["우편번호"].isna().sum()

print("\n재조회 전 우편번호 미확보:", missing_before)

if missing_before > 0:

    KAKAO_REST_KEY = getpass.getpass(
        "카카오 REST API 키를 입력하세요: "
    )

    url = (
        "https://dapi.kakao.com/v2/local/search/address.json"
    )

    headers = {
        "Authorization": f"KakaoAK {KAKAO_REST_KEY}"
    }

    for idx in warehouse[
        warehouse["우편번호"].isna()
    ].index:

        query = warehouse.loc[
            idx,
            "우편번호검색주소"
        ]

        try:
            response = requests.get(
                url,
                headers=headers,
                params={"query": query},
                timeout=20
            )

            response.raise_for_status()

            documents = response.json().get(
                "documents",
                []
            )

            if documents:

                result = documents[0]

                road_address = (
                    result.get("road_address")
                    or {}
                )

                zone_no = road_address.get(
                    "zone_no"
                )

                if zone_no:
                    warehouse.loc[
                        idx,
                        "우편번호"
                    ] = str(zone_no).zfill(5)

                    warehouse.loc[
                        idx,
                        "우편번호검색상태"
                    ] = "성공"

                else:
                    warehouse.loc[
                        idx,
                        "우편번호검색상태"
                    ] = "도로명주소_우편번호없음"

            else:
                warehouse.loc[
                    idx,
                    "우편번호검색상태"
                ] = "검색결과없음"

        except Exception as e:

            warehouse.loc[
                idx,
                "우편번호검색상태"
            ] = f"오류:{type(e).__name__}"

        time.sleep(0.1)

else:
    warehouse["우편번호검색상태"] = "기존값사용"


# ---------------------------------------------------------
# 5. 우편번호 최종 형식 정리
# ---------------------------------------------------------

warehouse["우편번호"] = (
    warehouse["우편번호"]
    .astype("string")
    .str.replace(r"\D", "", regex=True)
    .str.zfill(5)
)

warehouse.loc[
    warehouse["우편번호"].str.len() != 5,
    "우편번호"
] = pd.NA


# ---------------------------------------------------------
# 6. 택배물량 파일의 우편번호와 실제 연결 가능한지 확인
# ---------------------------------------------------------

parcel_path = (
    Path(
        "/content/drive/MyDrive/대전_물류창고_폭염위험도/01_원본데이터"
    )
    / "과학기술정보통신부 우정사업본부_전국 우편번호(배달지 기준)별 택배물량_20231231.csv"
)

parcel = pd.read_csv(
    parcel_path,
    encoding="cp949",
    dtype={"우편번호": str}
)

parcel["우편번호"] = (
    parcel["우편번호"]
    .astype("string")
    .str.replace(r"\D", "", regex=True)
    .str.zfill(5)
)

parcel_zip_set = set(
    parcel["우편번호"].dropna()
)

warehouse["택배물량_우편번호일치"] = (
    warehouse["우편번호"]
    .isin(parcel_zip_set)
)


# ---------------------------------------------------------
# 7. 최소 검수
# ---------------------------------------------------------

print("\n[물류창고 우편번호 확보 결과]")

print(
    "전체 창고 수:",
    len(warehouse)
)

print(
    "우편번호 확보:",
    warehouse["우편번호"].notna().sum()
)

print(
    "우편번호 미확보:",
    warehouse["우편번호"].isna().sum()
)

print(
    "고유 우편번호 수:",
    warehouse["우편번호"].nunique()
)

print(
    "택배물량 자료와 일치:",
    warehouse["택배물량_우편번호일치"].sum()
)

print(
    "택배물량 자료와 불일치:",
    (~warehouse["택배물량_우편번호일치"]).sum()
)

display(
    warehouse[
        [
            "WARE_NO",
            "COMPANY_NAME",
            "COMPANY_ADDRESS",
            "우편번호",
            "우편번호검색상태",
            "택배물량_우편번호일치"
        ]
    ]
)


# ---------------------------------------------------------
# 8. 저장
# ---------------------------------------------------------

save_path = (
    PROCESSED_DIR /
    "대전_43개물류창고_우편번호_확보.csv"
)

warehouse.to_csv(
    save_path,
    index=False,
    encoding="utf-8-sig"
)

print("\n저장 완료:")
print(save_path)
```
