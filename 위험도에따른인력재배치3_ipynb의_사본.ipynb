```python
# =========================================================
# 1-2. 최종 기상·물류 데이터 구조 통합 검수
# =========================================================

import pandas as pd

# ---------------------------------------------------------
# 1. 파일 불러오기
# ---------------------------------------------------------

locations = pd.read_csv(
    location_path,
    encoding="utf-8-sig",
    dtype={
        "분석거점ID": str,
        "우편번호": str
    }
)

forecast = pd.read_csv(
    forecast_path,
    encoding="utf-8-sig",
    dtype={"WARE_NO": str}
)

parcel = pd.read_csv(
    parcel_path,
    encoding="cp949",
    low_memory=False
)

monthly = pd.read_csv(
    monthly_path,
    encoding="utf-8-sig"
)


# ---------------------------------------------------------
# 2. 공통 구조 출력 함수
# ---------------------------------------------------------

def inspect_dataframe(name, df):

    print("\n" + "=" * 70)
    print(f"[{name}]")
    print("=" * 70)

    print("행 수:", len(df))
    print("열 수:", len(df.columns))
    print("열 이름:")
    print(df.columns.tolist())

    print("\n전체 중복 행 수:")
    print(df.duplicated().sum())

    print("\n결측치가 있는 열:")
    missing = df.isna().sum()
    missing = missing[missing > 0].sort_values(
        ascending=False
    )

    if len(missing) == 0:
        print("없음")
    else:
        print(missing.head(20))

    print("\n앞부분 확인:")
    display(df.head(3))


# ---------------------------------------------------------
# 3. 파일별 기본 구조 검수
# ---------------------------------------------------------

inspect_dataframe(
    "최종 39개 물류거점",
    locations
)

inspect_dataframe(
    "시간별 체감온도 단기예보",
    forecast
)

inspect_dataframe(
    "우편번호별 택배물량",
    parcel
)

inspect_dataframe(
    "1개월 전망",
    monthly
)


# ---------------------------------------------------------
# 4. 단기예보 핵심 시간열 자동 확인
# ---------------------------------------------------------

forecast_time_candidates = [
    "예보일시",
    "예측일시",
    "forecast_datetime",
    "일시",
    "datetime"
]

forecast_time_col = next(
    (
        col
        for col in forecast_time_candidates
        if col in forecast.columns
    ),
    None
)

print("\n" + "=" * 70)
print("[단기예보 시간 범위 검수]")
print("=" * 70)

print("확인된 시간 열:", forecast_time_col)

if forecast_time_col is not None:

    forecast[forecast_time_col] = pd.to_datetime(
        forecast[forecast_time_col],
        errors="coerce"
    )

    print(
        "시간 변환 실패:",
        forecast[forecast_time_col].isna().sum()
    )

    print(
        "최초 예보시각:",
        forecast[forecast_time_col].min()
    )

    print(
        "최종 예보시각:",
        forecast[forecast_time_col].max()
    )

    print(
        "고유 예보시각 수:",
        forecast[forecast_time_col].nunique()
    )


# ---------------------------------------------------------
# 5. 단기예보 창고·격자 연결 구조 확인
# ---------------------------------------------------------

print("\n" + "=" * 70)
print("[단기예보 연결키 검수]")
print("=" * 70)

for col in [
    "WARE_NO",
    "nx",
    "ny",
    "격자X",
    "격자Y"
]:

    if col in forecast.columns:

        print(
            f"{col} 고유값 수:",
            forecast[col].nunique(
                dropna=True
            )
        )

        print(
            f"{col} 결측:",
            forecast[col].isna().sum()
        )


# ---------------------------------------------------------
# 6. 체감온도 핵심 열 후보 확인
# ---------------------------------------------------------

heat_candidates = [
    "체감온도",
    "기상청_체감온도",
    "체감온도_공식",
    "feels_like"
]

temperature_candidates = [
    "기온",
    "TMP",
    "temperature"
]

humidity_candidates = [
    "습도",
    "REH",
    "humidity"
]

heat_col = next(
    (
        col
        for col in heat_candidates
        if col in forecast.columns
    ),
    None
)

temp_col = next(
    (
        col
        for col in temperature_candidates
        if col in forecast.columns
    ),
    None
)

humidity_col = next(
    (
        col
        for col in humidity_candidates
        if col in forecast.columns
    ),
    None
)

print("\n" + "=" * 70)
print("[단기예보 핵심 기상 열 확인]")
print("=" * 70)

print("체감온도 열:", heat_col)
print("기온 열:", temp_col)
print("습도 열:", humidity_col)

for label, col in [
    ("체감온도", heat_col),
    ("기온", temp_col),
    ("습도", humidity_col)
]:

    if col is not None:

        numeric = pd.to_numeric(
            forecast[col],
            errors="coerce"
        )

        print(
            f"\n{label} 결측:",
            numeric.isna().sum()
        )

        print(
            f"{label} 범위:",
            numeric.min(),
            "~",
            numeric.max()
        )


# ---------------------------------------------------------
# 7. 검수 완료 표시
# ---------------------------------------------------------

print("\n통합 전 데이터 구조 검수 완료")
```

2-5. 38개 거점과 우편번호별 택배물량 연결 검수

```python
# =========================================================
# 2-5. 38개 거점과 우편번호별 택배물량 연결 검수
# =========================================================

from pathlib import Path
import pandas as pd
import numpy as np

PROCESSED_DIR = Path(
    "/content/drive/MyDrive/대전_물류창고_폭염위험도/03_전처리데이터"
)

# ---------------------------------------------------------
# 1. 38개 최종 거점 파일 불러오기
# ---------------------------------------------------------

location_path = (
    PROCESSED_DIR /
    "대전_최종_물류거점_38개_좌표격자우편번호완료.csv"
)

locations_38 = pd.read_csv(
    location_path,
    encoding="utf-8-sig",
    dtype={
        "분석거점ID": str,
        "우편번호": str
    }
)

# 우편번호를 5자리 문자열로 통일
locations_38["우편번호"] = (
    locations_38["우편번호"]
    .astype("string")
    .str.replace(r"\.0$", "", regex=True)
    .str.zfill(5)
)


# ---------------------------------------------------------
# 2. 우편번호별 택배물량 형식 정리
# ---------------------------------------------------------

parcel_clean = parcel.copy()

parcel_clean["우편번호"] = (
    parcel_clean["우편번호"]
    .astype("string")
    .str.replace(r"\.0$", "", regex=True)
    .str.zfill(5)
)

month_cols = [
    "1월", "2월", "3월", "4월",
    "5월", "6월", "7월", "8월",
    "9월", "10월", "11월", "12월"
]

for col in month_cols:

    parcel_clean[col] = pd.to_numeric(
        parcel_clean[col],
        errors="coerce"
    )


# ---------------------------------------------------------
# 3. 우편번호별 물량 자료 기본 검수
# ---------------------------------------------------------

print("[택배물량 원자료 검수]")

print(
    "전체 우편번호 수:",
    parcel_clean["우편번호"].nunique()
)

print(
    "우편번호 중복:",
    parcel_clean["우편번호"].duplicated().sum()
)

print(
    "월별 물량 결측 행:",
    parcel_clean[month_cols].isna().any(axis=1).sum()
)


# ---------------------------------------------------------
# 4. 38개 거점에 우편번호별 택배물량 연결
# ---------------------------------------------------------

location_parcel = locations_38.merge(
    parcel_clean[
        ["우편번호"] + month_cols
    ],
    on="우편번호",
    how="left",
    validate="many_to_one"
)


# ---------------------------------------------------------
# 5. 같은 우편번호를 공유하는 거점 수 계산
# ---------------------------------------------------------

zip_location_count = (
    location_parcel
    .groupby(
        "우편번호"
    )["분석거점ID"]
    .transform("nunique")
)

location_parcel[
    "동일우편번호_거점수"
] = zip_location_count


# ---------------------------------------------------------
# 6. 여름철 물량 파생변수 생성
#
# 6~9월 합계와 월평균을 우선 생성
# 아직 거점별 배분은 하지 않음
# ---------------------------------------------------------

summer_cols = [
    "6월",
    "7월",
    "8월",
    "9월"
]

location_parcel[
    "여름철_택배물량합계"
] = location_parcel[
    summer_cols
].sum(axis=1)

location_parcel[
    "여름철_월평균택배물량"
] = location_parcel[
    summer_cols
].mean(axis=1)


# ---------------------------------------------------------
# 7. 연결 결과 검수
# ---------------------------------------------------------

print("\n[38개 거점 택배물량 연결 결과]")

print(
    "최종 거점 수:",
    len(location_parcel)
)

print(
    "택배물량 미연결 거점:",
    location_parcel[
        month_cols
    ].isna().all(axis=1).sum()
)

print(
    "연결된 고유 우편번호 수:",
    location_parcel[
        "우편번호"
    ].nunique()
)

print(
    "같은 우편번호를 공유하는 거점 수:",
    (
        location_parcel[
            "동일우편번호_거점수"
        ] > 1
    ).sum()
)

print(
    "복수 거점이 존재하는 우편번호 수:",
    location_parcel.loc[
        location_parcel[
            "동일우편번호_거점수"
        ] > 1,
        "우편번호"
    ].nunique()
)

print(
    "여름철 물량 범위:",
    int(
        location_parcel[
            "여름철_택배물량합계"
        ].min()
    ),
    "~",
    int(
        location_parcel[
            "여름철_택배물량합계"
        ].max()
    )
)


# ---------------------------------------------------------
# 8. 동일 우편번호 공유 거점 출력
# ---------------------------------------------------------

shared_zip = (
    location_parcel[
        location_parcel[
            "동일우편번호_거점수"
        ] > 1
    ]
    .sort_values(
        [
            "우편번호",
            "대표거점명"
        ]
    )
)

print("\n[동일 우편번호 공유 거점]")

if len(shared_zip) == 0:

    print("없음")

else:

    display(
        shared_zip[
            [
                "분석거점ID",
                "대표거점명",
                "대표주소",
                "우편번호",
                "동일우편번호_거점수",
                "대표_보관면적",
                "6월",
                "7월",
                "8월",
                "9월",
                "여름철_택배물량합계"
            ]
        ]
    )


# ---------------------------------------------------------
# 9. 물량 미연결 거점이 있으면 출력
# ---------------------------------------------------------

unmatched = location_parcel[
    location_parcel[
        month_cols
    ].isna().all(axis=1)
]

print("\n[택배물량 미연결 거점]")

if len(unmatched) == 0:

    print("없음")

else:

    display(
        unmatched[
            [
                "분석거점ID",
                "대표거점명",
                "대표주소",
                "우편번호"
            ]
        ]
    )


# ---------------------------------------------------------
# 10. 연결 결과 저장
# ---------------------------------------------------------

save_path = (
    PROCESSED_DIR /
    "대전_38개물류거점_우편번호별택배물량연결.csv"
)

location_parcel.to_csv(
    save_path,
    index=False,
    encoding="utf-8-sig"
)

print("\n저장 완료:")
print(save_path)
```

2-9. 시간별 기상예보와 거점별 하루 물류부하 결함

```python
# @title
# =========================================================
# 2-9. 시간별 기상예보와 거점별 하루 물류부하 결합
#
# 분석 방향:
# 시간대별 실제 기상예보 + 거점별 하루 평균 물류부하
#
# 시간대별 물류량은 임의로 추정하지 않음
# =========================================================

from pathlib import Path
import pandas as pd
import numpy as np

PROCESSED_DIR = Path(
    "/content/drive/MyDrive/대전_물류창고_폭염위험도/03_전처리데이터"
)


# ---------------------------------------------------------
# 1. 38개 거점 시간별 기상예보 불러오기
# ---------------------------------------------------------

forecast_path = (
    PROCESSED_DIR /
    "대전_38개물류거점_시간별체감온도예보.csv"
)

forecast_38 = pd.read_csv(
    forecast_path,
    encoding="utf-8-sig",
    dtype={
        "분석거점ID": str,
        "우편번호": str
    }
)

forecast_38["예보일시"] = pd.to_datetime(
    forecast_38["예보일시"],
    errors="coerce"
)


# ---------------------------------------------------------
# 2. 거점별 하루 평균 물류부하 불러오기
# ---------------------------------------------------------

daily_load_path = (
    PROCESSED_DIR /
    "대전_38개물류거점_7월일평균물류부하.csv"
)

daily_load = pd.read_csv(
    daily_load_path,
    encoding="utf-8-sig",
    dtype={
        "분석거점ID": str,
        "우편번호": str
    }
)

daily_load["7월_거점일평균물류부하"] = pd.to_numeric(
    daily_load["7월_거점일평균물류부하"],
    errors="coerce"
)

daily_load["거점상대물류부하지수"] = pd.to_numeric(
    daily_load["거점상대물류부하지수"],
    errors="coerce"
)

daily_load["거점물류부하백분위"] = pd.to_numeric(
    daily_load["거점물류부하백분위"],
    errors="coerce"
)


# ---------------------------------------------------------
# 3. 결합에 필요한 물류부하 열만 선택
#
# 기상예보 파일에 이미 거점명·주소·좌표가 있으므로
# 중복 열이 생기지 않도록 물류부하 관련 열만 가져옴
# ---------------------------------------------------------

load_cols = [
    "분석거점ID",
    "물량배분방식",
    "물량배분비율",
    "7월_거점배분물량",
    "7월_거점일평균물류부하",
    "거점상대물류부하지수",
    "거점물류부하백분위"
]

daily_load_for_merge = daily_load[
    load_cols
].copy()


# ---------------------------------------------------------
# 4. 시간별 기상예보에 거점별 하루 물류부하 연결
#
# 같은 거점의 하루 물류부하는 모든 시간대에 동일하게 연결됨
# 시간에 따라 변하는 값: TMP, REH, 체감온도
# 거점별로 고정되는 값: 하루 물류부하
# ---------------------------------------------------------

risk_base = forecast_38.merge(
    daily_load_for_merge,
    on="분석거점ID",
    how="left",
    validate="many_to_one"
)


# ---------------------------------------------------------
# 5. 날짜·시간 파생변수 생성
# ---------------------------------------------------------

risk_base["예보날짜"] = (
    risk_base["예보일시"].dt.date
)

risk_base["예보시간"] = (
    risk_base["예보일시"].dt.hour
)

risk_base["요일"] = (
    risk_base["예보일시"]
    .dt.day_name()
    .map({
        "Monday": "월",
        "Tuesday": "화",
        "Wednesday": "수",
        "Thursday": "목",
        "Friday": "금",
        "Saturday": "토",
        "Sunday": "일"
    })
)


# ---------------------------------------------------------
# 6. 분석 구조 검수
# ---------------------------------------------------------

expected_rows = (
    forecast_38["분석거점ID"].nunique()
    * forecast_38["예보일시"].nunique()
)

print("[시간별 기상예보 + 하루 물류부하 결합 결과]")

print(
    "예상 행 수:",
    expected_rows
)

print(
    "실제 행 수:",
    len(risk_base)
)

print(
    "최종 거점 수:",
    risk_base["분석거점ID"].nunique()
)

print(
    "고유 예보시각 수:",
    risk_base["예보일시"].nunique()
)

print(
    "거점·예보시각 중복:",
    risk_base[
        ["분석거점ID", "예보일시"]
    ].duplicated().sum()
)

print(
    "예보일시 결측:",
    risk_base["예보일시"].isna().sum()
)

print(
    "핵심 기상값 결측:",
    risk_base[
        ["TMP", "REH", "체감온도"]
    ].isna().any(axis=1).sum()
)

print(
    "하루 물류부하 미연결:",
    risk_base[
        "7월_거점일평균물류부하"
    ].isna().sum()
)

print(
    "상대물류부하지수 미연결:",
    risk_base[
        "거점상대물류부하지수"
    ].isna().sum()
)

print(
    "예보 기간:",
    risk_base["예보일시"].min(),
    "~",
    risk_base["예보일시"].max()
)

print(
    "체감온도 범위:",
    round(risk_base["체감온도"].min(), 1),
    "~",
    round(risk_base["체감온도"].max(), 1)
)

print(
    "거점별 하루 물류부하 범위:",
    round(
        risk_base[
            "7월_거점일평균물류부하"
        ].min(),
        3
    ),
    "~",
    round(
        risk_base[
            "7월_거점일평균물류부하"
        ].max(),
        3
    )
)


# ---------------------------------------------------------
# 7. 한 거점에서 시간별 기상값과
#    고정된 하루 물류부하가 함께 연결됐는지 확인
# ---------------------------------------------------------

sample_location_id = (
    daily_load
    .sort_values(
        "7월_거점일평균물류부하",
        ascending=False
    )
    .iloc[0]["분석거점ID"]
)

print("\n[물류부하 상위 거점 결합 예시]")

display(
    risk_base[
        risk_base["분석거점ID"]
        == sample_location_id
    ][
        [
            "분석거점ID",
            "대표거점명",
            "예보일시",
            "TMP",
            "REH",
            "체감온도",
            "7월_거점일평균물류부하",
            "거점상대물류부하지수",
            "거점물류부하백분위"
        ]
    ].head(15)
)


# ---------------------------------------------------------
# 8. 거점별 물류부하가 시간에 따라
#    잘못 달라지지 않는지 검수
# ---------------------------------------------------------

load_consistency = (
    risk_base
    .groupby(
        "분석거점ID"
    )[
        "7월_거점일평균물류부하"
    ]
    .nunique()
)

print("\n[거점별 하루 물류부하 일관성 검수]")

print(
    "시간에 따라 물류부하가 달라진 거점 수:",
    (load_consistency > 1).sum()
)


# ---------------------------------------------------------
# 9. 저장
# ---------------------------------------------------------

save_path = (
    PROCESSED_DIR /
    "대전_38개물류거점_시간별기상_하루물류부하결합.csv"
)

risk_base.to_csv(
    save_path,
    index=False,
    encoding="utf-8-sig"
)

print("\n저장 완료:")
print(save_path)
```

```python
# =========================================================
# 3-1. 체감온도 기준 시간별 작업 기상위험 등급 설정
#
# 주의:
# 이 등급은 기상청 공식 폭염특보를 직접 판정하는 것이 아니라,
# 사업장 폭염 대응기준을 참고한 시간별 작업위험 분류임
#
# 33℃ 미만      : 낮음
# 33~35℃ 미만   : 주의
# 35~38℃ 미만   : 경고
# 38℃ 이상      : 매우 위험
# =========================================================

from pathlib import Path
import pandas as pd
import numpy as np

PROCESSED_DIR = Path(
    "/content/drive/MyDrive/대전_물류창고_폭염위험도/03_전처리데이터"
)


# ---------------------------------------------------------
# 1. 2-9 결합 결과 불러오기
# ---------------------------------------------------------

risk_base_path = (
    PROCESSED_DIR /
    "대전_38개물류거점_시간별기상_하루물류부하결합.csv"
)

risk_weather = pd.read_csv(
    risk_base_path,
    encoding="utf-8-sig",
    dtype={
        "분석거점ID": str,
        "우편번호": str
    }
)

risk_weather["예보일시"] = pd.to_datetime(
    risk_weather["예보일시"],
    errors="coerce"
)

risk_weather["체감온도"] = pd.to_numeric(
    risk_weather["체감온도"],
    errors="coerce"
)


# ---------------------------------------------------------
# 2. 체감온도 기준 시간별 작업 기상위험 등급 설정
#
# 33℃ 미만      : 낮음
# 33~35℃ 미만   : 주의
# 35~38℃ 미만   : 경고
# 38℃ 이상      : 매우 위험
# ---------------------------------------------------------

temperature_bins = [
    -np.inf,
    33,
    35,
    38,
    np.inf
]

weather_labels = [
    "낮음",
    "주의",
    "경고",
    "매우 위험"
]

risk_weather["기상위험등급"] = pd.cut(
    risk_weather["체감온도"],
    bins=temperature_bins,
    labels=weather_labels,
    right=False,
    ordered=True
)


# ---------------------------------------------------------
# 3. 기상위험점수 부여
#
# 낮음      : 1점
# 주의      : 2점
# 경고      : 3점
# 매우 위험 : 4점
# ---------------------------------------------------------

weather_score_map = {
    "낮음": 1,
    "주의": 2,
    "경고": 3,
    "매우 위험": 4
}

risk_weather["기상위험점수"] = (
    risk_weather["기상위험등급"]
    .astype("string")
    .map(weather_score_map)
    .astype("Int64")
)


# ---------------------------------------------------------
# 4. 날짜·시간 정보 생성
# ---------------------------------------------------------

risk_weather["예보날짜"] = (
    risk_weather["예보일시"].dt.date
)

risk_weather["예보시간"] = (
    risk_weather["예보일시"].dt.hour
)


# ---------------------------------------------------------
# 5. 기본 검수
# ---------------------------------------------------------

print("[시간별 작업 기상위험 등급 산출 결과]")

print(
    "전체 행 수:",
    len(risk_weather)
)

print(
    "최종 거점 수:",
    risk_weather["분석거점ID"].nunique()
)

print(
    "고유 예보시각 수:",
    risk_weather["예보일시"].nunique()
)

print(
    "예보일시 결측:",
    risk_weather["예보일시"].isna().sum()
)

print(
    "체감온도 결측:",
    risk_weather["체감온도"].isna().sum()
)

print(
    "기상위험등급 결측:",
    risk_weather["기상위험등급"].isna().sum()
)

print(
    "기상위험점수 결측:",
    risk_weather["기상위험점수"].isna().sum()
)

print(
    "체감온도 범위:",
    round(risk_weather["체감온도"].min(), 1),
    "~",
    round(risk_weather["체감온도"].max(), 1)
)


# ---------------------------------------------------------
# 6. 기상위험등급별 행 수와 비율
# ---------------------------------------------------------

weather_summary = (
    risk_weather["기상위험등급"]
    .value_counts(
        sort=False,
        dropna=False
    )
    .rename_axis("기상위험등급")
    .reset_index(name="행수")
)

weather_summary["비율_pct"] = (
    weather_summary["행수"]
    / len(risk_weather)
    * 100
).round(2)

print("\n[기상위험등급별 분포]")

display(weather_summary)


# ---------------------------------------------------------
# 7. 예보시각별 기상위험 요약
#
# 각 시간에 대전 38개 거점 중:
# - 최고 체감온도
# - 평균 체감온도
# - 가장 높은 위험등급
# - 주의 이상 거점 수
# - 경고 이상 거점 수
# 를 계산함
# ---------------------------------------------------------

hourly_weather_summary = (
    risk_weather
    .groupby(
        "예보일시",
        as_index=False
    )
    .agg(
        최고체감온도=(
            "체감온도",
            "max"
        ),
        평균체감온도=(
            "체감온도",
            "mean"
        ),
        최고기상위험점수=(
            "기상위험점수",
            "max"
        ),
        주의이상_거점수=(
            "기상위험점수",
            lambda x: (x >= 2).sum()
        ),
        경고이상_거점수=(
            "기상위험점수",
            lambda x: (x >= 3).sum()
        ),
        매우위험_거점수=(
            "기상위험점수",
            lambda x: (x >= 4).sum()
        )
    )
)

score_to_grade = {
    1: "낮음",
    2: "주의",
    3: "경고",
    4: "매우 위험"
}

hourly_weather_summary["최고기상위험등급"] = (
    hourly_weather_summary["최고기상위험점수"]
    .map(score_to_grade)
)


# ---------------------------------------------------------
# 8. 위험도가 높은 예보시각 확인
# ---------------------------------------------------------

print("\n[최고 체감온도 상위 15개 예보시각]")

display(
    hourly_weather_summary
    .sort_values(
        [
            "최고기상위험점수",
            "최고체감온도"
        ],
        ascending=False
    )
    .head(15)
)


# ---------------------------------------------------------
# 9. 날짜별 기상위험 요약
#
# 주의이상_거점시간수:
# 하루 동안 38개 거점에서 발생한
# '주의 이상' 시간별 기록의 총개수
# ---------------------------------------------------------

daily_weather_summary = (
    risk_weather
    .groupby(
        "예보날짜",
        as_index=False
    )
    .agg(
        일최고체감온도=(
            "체감온도",
            "max"
        ),
        일평균체감온도=(
            "체감온도",
            "mean"
        ),
        일최고기상위험점수=(
            "기상위험점수",
            "max"
        ),
        주의이상_거점시간수=(
            "기상위험점수",
            lambda x: (x >= 2).sum()
        ),
        경고이상_거점시간수=(
            "기상위험점수",
            lambda x: (x >= 3).sum()
        ),
        매우위험_거점시간수=(
            "기상위험점수",
            lambda x: (x >= 4).sum()
        )
    )
)

daily_weather_summary["일최고기상위험등급"] = (
    daily_weather_summary["일최고기상위험점수"]
    .map(score_to_grade)
)

print("\n[날짜별 기상위험 요약]")

display(daily_weather_summary)


# ---------------------------------------------------------
# 10. 등급 경계값이 정확히 적용됐는지 검수
# ---------------------------------------------------------

boundary_check = pd.DataFrame({
    "체감온도_예시": [
        32.9,
        33.0,
        34.9,
        35.0,
        37.9,
        38.0
    ]
})

boundary_check["적용등급"] = pd.cut(
    boundary_check["체감온도_예시"],
    bins=temperature_bins,
    labels=weather_labels,
    right=False,
    ordered=True
)

print("\n[체감온도 경계값 적용 확인]")

display(boundary_check)


# ---------------------------------------------------------
# 11. 저장
# ---------------------------------------------------------

risk_path = (
    PROCESSED_DIR /
    "대전_38개물류거점_시간별작업기상위험등급.csv"
)

hourly_path = (
    PROCESSED_DIR /
    "대전_예보시각별_작업기상위험요약.csv"
)

daily_path = (
    PROCESSED_DIR /
    "대전_날짜별_작업기상위험요약.csv"
)

risk_weather.to_csv(
    risk_path,
    index=False,
    encoding="utf-8-sig"
)

hourly_weather_summary.to_csv(
    hourly_path,
    index=False,
    encoding="utf-8-sig"
)

daily_weather_summary.to_csv(
    daily_path,
    index=False,
    encoding="utf-8-sig"
)

print("\n저장 완료:")
print(risk_path)
print(hourly_path)
print(daily_path)
```

3-2. 거점별 하루 물류부하 등급 설정

```python
# =========================================================
# 3-2. 거점별 하루 물류부하 등급 설정
#
# 목적:
# 우편번호별 택배수요를 배분하여 산출한
# 거점별 하루 평균 물류부하를 상대적 백분위로 구분함
#
# 50백분위 미만       : 낮음
# 50~75백분위 미만    : 보통
# 75백분위 이상       : 높음
#
# 주의:
# 실제 물류창고 처리량이 아니라
# 거점별 상대적 물류부하 대리변수임
# =========================================================

from pathlib import Path
import pandas as pd
import numpy as np

PROCESSED_DIR = Path(
    "/content/drive/MyDrive/대전_물류창고_폭염위험도/03_전처리데이터"
)


# ---------------------------------------------------------
# 1. 거점별 하루 평균 물류부하 자료 불러오기
# ---------------------------------------------------------

daily_load_path = (
    PROCESSED_DIR /
    "대전_38개물류거점_7월일평균물류부하.csv"
)

load_grade = pd.read_csv(
    daily_load_path,
    encoding="utf-8-sig",
    dtype={
        "분석거점ID": str,
        "우편번호": str
    }
)


# ---------------------------------------------------------
# 2. 분석에 필요한 숫자형 열 변환
# ---------------------------------------------------------

numeric_cols = [
    "7월_거점배분물량",
    "7월_거점일평균물류부하",
    "거점상대물류부하지수",
    "거점물류부하백분위"
]

for col in numeric_cols:
    if col in load_grade.columns:
        load_grade[col] = pd.to_numeric(
            load_grade[col],
            errors="coerce"
        )


# ---------------------------------------------------------
# 3. 필수 열 존재 여부 확인
# ---------------------------------------------------------

required_cols = [
    "분석거점ID",
    "7월_거점일평균물류부하",
    "거점물류부하백분위"
]

missing_cols = [
    col for col in required_cols
    if col not in load_grade.columns
]

if missing_cols:
    raise KeyError(
        "필수 열이 없습니다: "
        + ", ".join(missing_cols)
    )


# ---------------------------------------------------------
# 4. 백분위 기준 물류부하 등급 설정
#
# 동일한 백분위값은 같은 등급으로 분류되므로,
# 같은 물류부하를 가진 거점이 서로 다른 등급으로
# 나뉘지 않도록 함
# ---------------------------------------------------------

conditions = [
    load_grade["거점물류부하백분위"] < 50,

    (
        (load_grade["거점물류부하백분위"] >= 50)
        & (load_grade["거점물류부하백분위"] < 75)
    ),

    load_grade["거점물류부하백분위"] >= 75
]

load_grade["물류부하등급"] = np.select(
    conditions,
    [
        "낮음",
        "보통",
        "높음"
    ],
    default=pd.NA
)


# ---------------------------------------------------------
# 5. 물류부하 점수 부여
#
# 낮음 : 1점
# 보통 : 2점
# 높음 : 3점
# ---------------------------------------------------------

load_score_map = {
    "낮음": 1,
    "보통": 2,
    "높음": 3
}

load_grade["물류부하점수"] = (
    load_grade["물류부하등급"]
    .map(load_score_map)
    .astype("Int64")
)


# ---------------------------------------------------------
# 6. 시각화용 등급 순서 설정
# ---------------------------------------------------------

load_order = [
    "낮음",
    "보통",
    "높음"
]

load_grade["물류부하등급"] = pd.Categorical(
    load_grade["물류부하등급"],
    categories=load_order,
    ordered=True
)


# ---------------------------------------------------------
# 7. 기본 검수
# ---------------------------------------------------------

print("[거점별 하루 물류부하 등급 산출 결과]")

print(
    "전체 행 수:",
    len(load_grade)
)

print(
    "고유 거점 수:",
    load_grade["분석거점ID"].nunique()
)

print(
    "분석거점ID 중복:",
    load_grade["분석거점ID"].duplicated().sum()
)

print(
    "하루 물류부하 결측:",
    load_grade[
        "7월_거점일평균물류부하"
    ].isna().sum()
)

print(
    "물류부하백분위 결측:",
    load_grade[
        "거점물류부하백분위"
    ].isna().sum()
)

print(
    "물류부하등급 결측:",
    load_grade[
        "물류부하등급"
    ].isna().sum()
)

print(
    "물류부하점수 결측:",
    load_grade[
        "물류부하점수"
    ].isna().sum()
)

print(
    "하루 물류부하 범위:",
    round(
        load_grade[
            "7월_거점일평균물류부하"
        ].min(),
        3
    ),
    "~",
    round(
        load_grade[
            "7월_거점일평균물류부하"
        ].max(),
        3
    )
)


# ---------------------------------------------------------
# 8. 물류부하 등급별 분포 요약
# ---------------------------------------------------------

load_summary = (
    load_grade
    .groupby(
        "물류부하등급",
        observed=False,
        as_index=False
    )
    .agg(
        거점수=(
            "분석거점ID",
            "nunique"
        ),
        최소_일평균물류부하=(
            "7월_거점일평균물류부하",
            "min"
        ),
        최대_일평균물류부하=(
            "7월_거점일평균물류부하",
            "max"
        ),
        최소_백분위=(
            "거점물류부하백분위",
            "min"
        ),
        최대_백분위=(
            "거점물류부하백분위",
            "max"
        )
    )
)

load_summary["거점비율_pct"] = (
    load_summary["거점수"]
    / load_grade["분석거점ID"].nunique()
    * 100
).round(2)

print("\n[물류부하등급별 분포]")

display(load_summary)


# ---------------------------------------------------------
# 9. 물류부하 상위 거점 확인
# ---------------------------------------------------------

display_cols = [
    col for col in [
        "분석거점ID",
        "대표거점명",
        "우편번호",
        "물량배분방식",
        "7월_거점일평균물류부하",
        "거점상대물류부하지수",
        "거점물류부하백분위",
        "물류부하등급",
        "물류부하점수"
    ]
    if col in load_grade.columns
]

print("\n[물류부하 상위 15개 거점]")

display(
    load_grade
    .sort_values(
        [
            "거점물류부하백분위",
            "7월_거점일평균물류부하"
        ],
        ascending=False
    )[display_cols]
    .head(15)
)


# ---------------------------------------------------------
# 10. 같은 물류부하가 서로 다른 등급으로
#     분리됐는지 검수
# ---------------------------------------------------------

tie_check = (
    load_grade
    .groupby(
        "7월_거점일평균물류부하",
        dropna=False
    )["물류부하등급"]
    .nunique()
)

split_tie_count = (
    tie_check > 1
).sum()

print(
    "\n동일 물류부하가 서로 다른 등급으로 "
    "분리된 값의 수:",
    split_tie_count
)


# ---------------------------------------------------------
# 11. 등급 경계값 적용 확인
# ---------------------------------------------------------

boundary_check = pd.DataFrame({
    "물류부하백분위_예시": [
        49.9,
        50.0,
        74.9,
        75.0,
        100.0
    ]
})

boundary_conditions = [
    boundary_check["물류부하백분위_예시"] < 50,

    (
        (boundary_check["물류부하백분위_예시"] >= 50)
        & (boundary_check["물류부하백분위_예시"] < 75)
    ),

    boundary_check["물류부하백분위_예시"] >= 75
]

boundary_check["적용등급"] = np.select(
    boundary_conditions,
    [
        "낮음",
        "보통",
        "높음"
    ],
    default=pd.NA
)

print("\n[물류부하 경계값 적용 확인]")

display(boundary_check)


# ---------------------------------------------------------
# 12. 저장
# ---------------------------------------------------------

save_path = (
    PROCESSED_DIR /
    "대전_38개물류거점_하루물류부하등급.csv"
)

summary_path = (
    PROCESSED_DIR /
    "대전_물류부하등급별_분포요약.csv"
)

load_grade.to_csv(
    save_path,
    index=False,
    encoding="utf-8-sig"
)

load_summary.to_csv(
    summary_path,
    index=False,
    encoding="utf-8-sig"
)

print("\n저장 완료:")
print(save_path)
print(summary_path)
```

3-3. 시간별 예보를 2시간 단위 위험 블록으로 변환

```python
# =========================================================
# 3-3. 시간별 예보를 2시간 단위 위험 블록으로 변환
#
# 목적:
# 시간별 체감온도 예보를 2시간 단위로 묶고,
# 각 블록에서 가장 높은 체감온도와 위험등급을 적용함
#
# 예:
# 12시와 13시 자료 → 12:00~14:00 블록
# 14시와 15시 자료 → 14:00~16:00 블록
#
# 근로자 안전을 고려하여 2시간 평균이 아니라
# 해당 블록의 최고 체감온도를 대표값으로 사용함
# =========================================================

from pathlib import Path
import pandas as pd
import numpy as np

PROCESSED_DIR = Path(
    "/content/drive/MyDrive/대전_물류창고_폭염위험도/03_전처리데이터"
)


# ---------------------------------------------------------
# 1. 시간별 작업 기상위험 자료 불러오기
# ---------------------------------------------------------

input_path = (
    PROCESSED_DIR /
    "대전_38개물류거점_시간별작업기상위험등급.csv"
)

hourly_risk = pd.read_csv(
    input_path,
    encoding="utf-8-sig",
    dtype={
        "분석거점ID": str,
        "우편번호": str
    }
)


# ---------------------------------------------------------
# 2. 날짜·시간 열 변환
# ---------------------------------------------------------

if "예보일시" not in hourly_risk.columns:
    raise KeyError("예보일시 열을 찾을 수 없습니다.")

hourly_risk["예보일시"] = pd.to_datetime(
    hourly_risk["예보일시"],
    errors="coerce"
)

if hourly_risk["예보일시"].isna().any():
    raise ValueError(
        "예보일시 변환 과정에서 결측이 발생했습니다."
    )


# ---------------------------------------------------------
# 3. 체감온도 열 확인
# ---------------------------------------------------------

apparent_candidates = [
    "체감온도",
    "예보체감온도",
    "시간별체감온도",
    "체감온도_C"
]

apparent_col = next(
    (
        col for col in apparent_candidates
        if col in hourly_risk.columns
    ),
    None
)

if apparent_col is None:
    raise KeyError(
        "체감온도 열을 찾을 수 없습니다. "
        f"현재 열 목록: {hourly_risk.columns.tolist()}"
    )

hourly_risk[apparent_col] = pd.to_numeric(
    hourly_risk[apparent_col],
    errors="coerce"
)

if hourly_risk[apparent_col].isna().any():
    raise ValueError(
        "체감온도 열에 숫자로 변환되지 않는 값이 있습니다."
    )


# ---------------------------------------------------------
# 4. 기상위험등급 점수 확인 및 생성
#
# 낮음       : 1
# 주의       : 2
# 경고       : 3
# 매우 위험  : 4
# ---------------------------------------------------------

weather_score_map = {
    "낮음": 1,
    "주의": 2,
    "경고": 3,
    "매우 위험": 4
}

if "기상위험등급" not in hourly_risk.columns:
    raise KeyError("기상위험등급 열을 찾을 수 없습니다.")

hourly_risk["기상위험점수"] = (
    hourly_risk["기상위험등급"]
    .map(weather_score_map)
    .astype("Int64")
)

if hourly_risk["기상위험점수"].isna().any():
    unknown_grade = (
        hourly_risk.loc[
            hourly_risk["기상위험점수"].isna(),
            "기상위험등급"
        ]
        .drop_duplicates()
        .tolist()
    )

    raise ValueError(
        "정의되지 않은 기상위험등급이 있습니다: "
        + str(unknown_grade)
    )


# ---------------------------------------------------------
# 5. 2시간 블록 시작시각 생성
#
# 예:
# 12:00, 13:00 → 12:00
# 14:00, 15:00 → 14:00
# ---------------------------------------------------------

hourly_risk["블록시작일시"] = (
    hourly_risk["예보일시"]
    .dt.floor("2h")
)

hourly_risk["블록종료일시"] = (
    hourly_risk["블록시작일시"]
    + pd.Timedelta(hours=2)
)


# ---------------------------------------------------------
# 6. 거점·2시간 블록별 자료 집계
#
# 대표 체감온도:
# 2시간 내 최고 체감온도
#
# 대표 위험등급:
# 2시간 내 가장 높은 기상위험점수
#
# 시간자료수:
# 정상적인 완전 블록은 2건
# 1건이면 부분 블록
# ---------------------------------------------------------

group_cols = [
    "분석거점ID",
    "블록시작일시",
    "블록종료일시"
]

agg_dict = {
    apparent_col: "max",
    "기상위험점수": "max",
    "예보일시": "count"
}

# 거점 정보 중 존재하는 열은 첫 번째 값을 유지
site_info_cols = [
    "대표거점명",
    "우편번호",
    "주소",
    "도로명주소",
    "격자X",
    "격자Y",
    "nx",
    "ny"
]

for col in site_info_cols:
    if col in hourly_risk.columns:
        agg_dict[col] = "first"

two_hour_risk = (
    hourly_risk
    .groupby(
        group_cols,
        as_index=False
    )
    .agg(agg_dict)
)


# ---------------------------------------------------------
# 7. 열 이름 정리
# ---------------------------------------------------------

two_hour_risk = two_hour_risk.rename(
    columns={
        apparent_col: "블록최고체감온도",
        "예보일시": "시간자료수"
    }
)

score_to_weather = {
    1: "낮음",
    2: "주의",
    3: "경고",
    4: "매우 위험"
}

two_hour_risk["블록기상위험등급"] = (
    two_hour_risk["기상위험점수"]
    .map(score_to_weather)
)

two_hour_risk["완전블록여부"] = np.where(
    two_hour_risk["시간자료수"] == 2,
    "완전",
    "부분"
)


# ---------------------------------------------------------
# 8. 날짜·근무조·표시용 시간대 생성
# ---------------------------------------------------------

two_hour_risk["기준날짜"] = (
    two_hour_risk["블록시작일시"]
    .dt.date
)

two_hour_risk["시작시각"] = (
    two_hour_risk["블록시작일시"]
    .dt.strftime("%H:%M")
)

two_hour_risk["종료시각"] = (
    two_hour_risk["블록종료일시"]
    .dt.strftime("%H:%M")
)

# 종료시각 00:00은 표에서 24:00으로 표시
midnight_mask = (
    two_hour_risk["종료시각"] == "00:00"
)

two_hour_risk.loc[
    midnight_mask,
    "종료시각"
] = "24:00"

two_hour_risk["시간대"] = (
    two_hour_risk["시작시각"]
    + "~"
    + two_hour_risk["종료시각"]
)

# 낮: 08:00~20:00
# 밤: 20:00~익일 08:00
start_hour = (
    two_hour_risk["블록시작일시"]
    .dt.hour
)

two_hour_risk["근무조"] = np.where(
    (start_hour >= 8) & (start_hour < 20),
    "낮",
    "밤"
)


# ---------------------------------------------------------
# 9. 열 순서 정리
# ---------------------------------------------------------

front_cols = [
    "기준날짜",
    "근무조",
    "시간대",
    "블록시작일시",
    "블록종료일시",
    "분석거점ID"
]

optional_front_cols = [
    col for col in [
        "대표거점명",
        "우편번호",
        "주소",
        "도로명주소",
        "격자X",
        "격자Y",
        "nx",
        "ny"
    ]
    if col in two_hour_risk.columns
]

result_cols = (
    front_cols
    + optional_front_cols
    + [
        "블록최고체감온도",
        "블록기상위험등급",
        "기상위험점수",
        "시간자료수",
        "완전블록여부"
    ]
)

two_hour_risk = two_hour_risk[result_cols]


# ---------------------------------------------------------
# 10. 기본 검수
# ---------------------------------------------------------

duplicate_count = two_hour_risk.duplicated(
    subset=[
        "분석거점ID",
        "블록시작일시"
    ]
).sum()

print("[2시간 단위 기상위험 블록 생성 결과]")

print(
    "전체 행 수:",
    len(two_hour_risk)
)

print(
    "고유 거점 수:",
    two_hour_risk["분석거점ID"].nunique()
)

print(
    "고유 2시간 블록 수:",
    two_hour_risk["블록시작일시"].nunique()
)

print(
    "거점·블록 중복:",
    duplicate_count
)

print(
    "블록최고체감온도 결측:",
    two_hour_risk[
        "블록최고체감온도"
    ].isna().sum()
)

print(
    "블록기상위험등급 결측:",
    two_hour_risk[
        "블록기상위험등급"
    ].isna().sum()
)

print(
    "체감온도 범위:",
    round(
        two_hour_risk[
            "블록최고체감온도"
        ].min(),
        1
    ),
    "~",
    round(
        two_hour_risk[
            "블록최고체감온도"
        ].max(),
        1
    )
)


# ---------------------------------------------------------
# 11. 완전·부분 블록 검수
# ---------------------------------------------------------

block_completion_summary = (
    two_hour_risk["완전블록여부"]
    .value_counts(dropna=False)
    .rename_axis("완전블록여부")
    .reset_index(name="거점블록수")
)

print("\n[완전·부분 블록 분포]")

display(block_completion_summary)


# ---------------------------------------------------------
# 12. 위험등급별 분포
# ---------------------------------------------------------

risk_order = [
    "낮음",
    "주의",
    "경고",
    "매우 위험"
]

two_hour_risk[
    "블록기상위험등급"
] = pd.Categorical(
    two_hour_risk["블록기상위험등급"],
    categories=risk_order,
    ordered=True
)

risk_summary = (
    two_hour_risk
    .groupby(
        "블록기상위험등급",
        observed=False,
        as_index=False
    )
    .agg(
        거점블록수=(
            "분석거점ID",
            "size"
        ),
        최소체감온도=(
            "블록최고체감온도",
            "min"
        ),
        최대체감온도=(
            "블록최고체감온도",
            "max"
        )
    )
)

risk_summary["비율_pct"] = (
    risk_summary["거점블록수"]
    / len(two_hour_risk)
    * 100
).round(2)

print("\n[2시간 블록 위험등급별 분포]")

display(risk_summary)


# ---------------------------------------------------------
# 13. 날짜·근무조별 블록 수 확인
# ---------------------------------------------------------

date_shift_summary = (
    two_hour_risk
    .groupby(
        [
            "기준날짜",
            "근무조",
            "완전블록여부"
        ],
        as_index=False,
        observed=False
    )
    .agg(
        고유시간블록수=(
            "블록시작일시",
            "nunique"
        ),
        거점블록수=(
            "분석거점ID",
            "size"
        )
    )
)

print("\n[날짜·근무조별 블록 수]")

display(date_shift_summary)


# ---------------------------------------------------------
# 14. 예시 거점 확인
#
# 물류부하 상위 거점 중 하나인 백년축산을 우선 표시
# 해당 거점이 없으면 첫 번째 거점을 표시
# ---------------------------------------------------------

if "대표거점명" in two_hour_risk.columns:
    sample_mask = (
        two_hour_risk["대표거점명"]
        .astype(str)
        .str.contains(
            "백년축산",
            na=False
        )
    )
else:
    sample_mask = pd.Series(
        False,
        index=two_hour_risk.index
    )

if sample_mask.any():
    sample_site = (
        two_hour_risk.loc[
            sample_mask,
            "분석거점ID"
        ]
        .iloc[0]
    )
else:
    sample_site = (
        two_hour_risk["분석거점ID"]
        .iloc[0]
    )

sample_cols = [
    col for col in [
        "기준날짜",
        "근무조",
        "시간대",
        "대표거점명",
        "블록최고체감온도",
        "블록기상위험등급",
        "시간자료수",
        "완전블록여부"
    ]
    if col in two_hour_risk.columns
]

print("\n[예시 거점의 2시간 단위 위험표]")

display(
    two_hour_risk.loc[
        two_hour_risk["분석거점ID"] == sample_site,
        sample_cols
    ]
    .sort_values("블록시작일시")
)


# ---------------------------------------------------------
# 15. 저장
# ---------------------------------------------------------

save_path = (
    PROCESSED_DIR /
    "대전_38개물류거점_2시간단위_작업기상위험.csv"
)

summary_path = (
    PROCESSED_DIR /
    "대전_2시간단위_작업기상위험등급별_요약.csv"
)

two_hour_risk.to_csv(
    save_path,
    index=False,
    encoding="utf-8-sig"
)

risk_summary.to_csv(
    summary_path,
    index=False,
    encoding="utf-8-sig"
)

print("\n저장 완료:")
print(save_path)
print(summary_path)
```

```python
# =========================================================
# 3-3-1. 예시 거점 표 출력 및 결과 저장 수정
#
# 수정 이유:
# 표시용 열을 먼저 선택한 뒤 블록시작일시로 정렬하면서
# 블록시작일시 열을 찾지 못하는 오류가 발생함
#
# 수정 방법:
# 블록시작일시를 기준으로 먼저 정렬한 뒤
# 최종 표시 열만 선택함
# =========================================================

from pathlib import Path
import pandas as pd

PROCESSED_DIR = Path(
    "/content/drive/MyDrive/대전_물류창고_폭염위험도/03_전처리데이터"
)


# ---------------------------------------------------------
# 1. 기존 3-3 실행 결과 존재 여부 확인
# ---------------------------------------------------------

required_variables = [
    "two_hour_risk",
    "risk_summary"
]

missing_variables = [
    var for var in required_variables
    if var not in globals()
]

if missing_variables:
    raise NameError(
        "기존 3-3 실행 결과가 메모리에 없습니다: "
        + ", ".join(missing_variables)
        + "\n런타임이 재시작된 경우에는 "
          "3-3 전체 코드를 다시 실행해야 합니다."
    )


# ---------------------------------------------------------
# 2. 예시 거점 선택
#
# 백년축산이 존재하면 백년축산을 표시하고,
# 없으면 첫 번째 거점을 표시함
# ---------------------------------------------------------

if "대표거점명" in two_hour_risk.columns:
    sample_mask = (
        two_hour_risk["대표거점명"]
        .astype(str)
        .str.contains(
            "백년축산",
            na=False
        )
    )
else:
    sample_mask = pd.Series(
        False,
        index=two_hour_risk.index
    )

if sample_mask.any():
    sample_site = (
        two_hour_risk.loc[
            sample_mask,
            "분석거점ID"
        ]
        .iloc[0]
    )
else:
    sample_site = (
        two_hour_risk["분석거점ID"]
        .iloc[0]
    )


# ---------------------------------------------------------
# 3. 예시 거점 자료를 먼저 시간순으로 정렬
# ---------------------------------------------------------

sample_result = (
    two_hour_risk.loc[
        two_hour_risk["분석거점ID"] == sample_site
    ]
    .sort_values(
        "블록시작일시"
    )
    .copy()
)


# ---------------------------------------------------------
# 4. 화면에 표시할 열 선택
# ---------------------------------------------------------

sample_cols = [
    col for col in [
        "기준날짜",
        "근무조",
        "시간대",
        "대표거점명",
        "블록최고체감온도",
        "블록기상위험등급",
        "시간자료수",
        "완전블록여부"
    ]
    if col in sample_result.columns
]

print("[예시 거점의 2시간 단위 위험표]")

display(
    sample_result[
        sample_cols
    ]
)


# ---------------------------------------------------------
# 5. 최종 저장 전 검수
# ---------------------------------------------------------

duplicate_count = two_hour_risk.duplicated(
    subset=[
        "분석거점ID",
        "블록시작일시"
    ]
).sum()

print("\n[저장 전 최종 검수]")

print(
    "전체 행 수:",
    len(two_hour_risk)
)

print(
    "고유 거점 수:",
    two_hour_risk["분석거점ID"].nunique()
)

print(
    "거점·블록 중복:",
    duplicate_count
)

print(
    "블록최고체감온도 결측:",
    two_hour_risk[
        "블록최고체감온도"
    ].isna().sum()
)

print(
    "블록기상위험등급 결측:",
    two_hour_risk[
        "블록기상위험등급"
    ].isna().sum()
)


# ---------------------------------------------------------
# 6. 결과 저장
# ---------------------------------------------------------

save_path = (
    PROCESSED_DIR /
    "대전_38개물류거점_2시간단위_작업기상위험.csv"
)

summary_path = (
    PROCESSED_DIR /
    "대전_2시간단위_작업기상위험등급별_요약.csv"
)

two_hour_risk.to_csv(
    save_path,
    index=False,
    encoding="utf-8-sig"
)

risk_summary.to_csv(
    summary_path,
    index=False,
    encoding="utf-8-sig"
)

print("\n저장 완료:")
print(save_path)
print(summary_path)
```
